# Data Analysis of our OFN dataset

```python
import pandas as pd

# Load the dataset
df = pd.read_csv('./data/nkod/nkod_ofn_metadata.csv')

# Number of samples
num_samples = len(df)

# Number of unique publishers
num_publishers = df['publisher_cs'].nunique()

# Mean number of keywords (splitting by ';')
df['kw_count'] = df['keywords_cs'].fillna('').apply(lambda x: len(x.split(';')) if x != '' else 0)
mean_keywords = df['kw_count'].mean()

# Mean length of description
df['desc_len'] = df['description_cs'].fillna('').apply(len)
mean_desc_len = df['desc_len'].mean()

print(f"Number of samples: {num_samples}")
print(f"Number of publishers: {num_publishers}")
print(f"Mean number of keywords: {mean_keywords:.2f}")
print(f"Mean length of description: {mean_desc_len:.2f}")
```
